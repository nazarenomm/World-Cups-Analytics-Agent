```python
import pandas as pd
```

```python
def filter_womens_world_cup(df):
    """
    Filters out rows from the DataFrame where the 'tournament_name' column contains "FIFA Women's World Cup".
    """
    return df[~df['tournament_name'].str.contains("FIFA Women's World Cup")]
```

## Award winners

```python
award_winners = pd.read_csv("data/raw/award_winners.csv")
```

```python
award_winners.head()
```

```python
award_winners = filter_womens_world_cup(award_winners)
```

```python
award_winners.head()
```

```python
award_winners.columns
```

```python
award_winners_proc = award_winners[['key_id', 'tournament_id', 'award_id', 'shared', 'player_id', 'team_id']]
award_winners_proc.rename(columns={'key_id': 'award_winner_id'}, inplace=True)
award_winners_proc
```

```python
award_winners_proc.to_csv("data/processed/award_winners.csv", index=False)
```
