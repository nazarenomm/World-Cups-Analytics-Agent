```python
import os
from pathlib import Path
from dotenv import load_dotenv
from wikipedia_fetch import search_wikipedia_candidates, preview_article, save_doc
```

```python
load_dotenv()

USER_AGENT = os.getenv("USER_AGENT")
OUTPUT_DIR = Path("./corpus/famous_matches")
OUTPUT_DIR.mkdir(exist_ok=True)
```

evento

```python
id = "battle_of_lusail_2022"
query = "Battle of Lusail 2022"
```

ingles

```python
candidates = search_wikipedia_candidates(query, lang="en")
for c in candidates:
    print(c["title"], "-", c["snippet"][:100])
```

```python
doc = preview_article(candidates[0]["title"], lang="en")
```

```python
save_doc(doc, event_id=id, output_dir=OUTPUT_DIR)
```

```python
doc_es = preview_article(candidates[0]["title"], lang="es")
```
